```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy.optimize as opt
from IPython.display import display, clear_output
import os
```

```python
def convModel(t, A, B, C):
    return C * (1 - A * np.exp(-B*t))

def cGd (Q, deltaT):
    m = 0.012583 # kg
    return Q / (m * deltaT)

def plotter (name, start, end):
    data = pd.read_csv(name).values
    time = data[:,1]
    sample = data[:,3]
    stage = data[:,2]
    ax.plot(time[start:end], sample[start:end], label='sample')
    ax.plot(time[start:end], stage[start:end], label='stage')
    ax.set_xlabel('time (s)')
    ax.set_ylabel('temperature (C)')
    ax.legend()
    disp.update(fig)

def cFinder(name, start, end):
    data = pd.read_csv(name).values
    time = data[:,1]
    sample = data[:,3]
    stage = data[:,2]
    P = 5**2 / 200
    timeDelta = time[end] - time[start]
    Q = P * timeDelta
    tempDelta = sample[end] - sample[start]
    c = cGd(Q, tempDelta)
    return c

def tempPlotter(c, name, start, end):
    data = pd.read_csv(name).values
    time = data[:,1]
    sample = data[:,3]
    P = 5**2 / 200
    m = 0.012583 # kg
    timearr = np.linspace(time[start], time[end], 1000)
    temps = (P / (m * c)) * (timearr - time[start]) + sample[start]
    ax.plot(timearr, temps,label='Linear Check Line')
    ax.legend()
    disp.update(fig)
```

```python
# defining some constants we will use throughout the analysis
V = 5 
R = 200
P = 5**2 / 200
offset = 0.25
start = 0
end = -1
```

```python
# constants 
looper = True

c = np.array([])

fig = plt.figure()
ax = fig.add_subplot(111)
disp = display(fig, display_id=True)
plt.close(fig)

while looper==True:
    cFount=0
    # constants
    start = 0
    end = -1


    
    #get function name and data
    name = input()
    data = pd.read_csv(name).values
    time = data[:,1]
    sample = data[:,3]
    stage = data[:,2]
    plotter(name, start, end)
    
    #time to start choosing data
    
    print(f'Analyzing {name} \nChoose a starting indice:')
    goodStart = 0
    chooser = 0
    goodEnd = 0
    
    # loop for function
    while chooser==0:
        cFound = 0
        ax.clear()
        start = int(input())
        plotter(name, start, end)
        print('Does this look good? 0 for no, 1 for yes')
        goodStart = int(input())
        if goodStart == 1:
            ax.clear()
            endarr = np.where(sample>(sample[start]+offset))[0]
            endloc = np.where(endarr > start)[0]
            end = endarr[endloc[0]]
            plotter(name, start, end)
            # finding c
            cFound = cFinder(name, start, end)
            tempPlotter(cFound, name, start, end)
            print('Does this look good? 0 for no, 1 for yes')
            goodEnd = int(input())   
        if goodEnd == 1:
            chooser = 1 

    print(f'The specific heat of Gadolinium is {cFound:.2f} J kg^-1 K^-1')
    c = np.append(c,cFound)
    
    # checking if done
    print('Done with loop? 0 for no, 1 for yes')
    done = int(input())
    print(done)
    if done==1:
        looper = False
        print('Done with loop, thank you')
    if done==0:
        ax.clear()

print('The specific heat of Gadolinium was found to be:\n',c)
```
